```python
#Ejercicio 4
res = np.sqrt(2)**2 - 2
print(res)

# 1. Definir 100 valores equiespaciados en el intervalo [0, 5e-8]
# Usamos float32 para notar la pérdida de precisión por cancelación numérica
x = np.linspace(0, 5e-8, 100, dtype=np.float32)

# 2. Evaluar ambas expresiones
y_a = np.sqrt(2 * x**2 + 1) - 1
y_b = (2 * x**2) / (np.sqrt(2 * x**2 + 1) + 1)

# 3. Graficar los resultados
plt.figure(figsize=(9, 5))
plt.plot(
    x, y_a, "ro--", label=r"a) $\sqrt{2x^2 + 1} - 1$ (Inestable)", alpha=0.7
)
plt.plot(
    x, y_b, "b.-", label=r"b) $\frac{2x^2}{\sqrt{2x^2+1}+1}$ (Estable)", alpha=0.7
)

plt.title("Comparación de Estabilidad Numérica")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Ejercicio 4
res = np.sqrt(2)**2 - 2
print(res)

# 1. Definir 100 valores equiespaciados en el intervalo [0, 5e-8]
# Usamos float32 para notar la pérdida de precisión por cancelación numérica
x = np.linspace(0, 5e-8, 100, dtype=np.float32)

# 2. Evaluar ambas expresiones
y_a = np.sqrt(2 * x**2 + 1) - 1
y_b = (2 * x**2) / (np.sqrt(2 * x**2 + 1) + 1)

# 3. Graficar los resultados
plt.figure(figsize=(9, 5))
plt.plot(
    x, y_a, "ro--", label=r"a) $\sqrt{2x^2 + 1} - 1$ (Inestable)", alpha=0.7
)
plt.plot(
    x, y_b, "b.-", label=r"b) $\frac{2x^2}{\sqrt{2x^2+1}+1}$ (Estable)", alpha=0.7
)

plt.title("Comparación de Estabilidad Numérica")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Ejercicio 5
import matplotlib.pyplot as plt
import math

# Inicialización de la lista con x1 = sqrt(2)
l = [math.sqrt(2)]

# Cálculo de los elementos x_i hasta 100 iteraciones
for i in range(1, 100):
    try:
        xi = (l[-1] ** 2) / math.sqrt(2)
        l.append(xi)
    except OverflowError:
        print(f"Desbordamiento numérico (Overflow) alcanzado en la iteración {i + 1}")
        break

# Graficar los valores
plt.figure(figsize=(10, 5))
plt.plot(range(1, len(l) + 1), l, marker='o', linestyle='-', markersize=3, color='b')
plt.title("Acumulación de Error en Sucesión Numérica")
plt.xlabel("Índice (i)")
plt.ylabel("Valor de $x_i$")
plt.grid(True)
plt.show()
```

```python
#Modulo ALC Ej 1
def error(x,y):
    x_f64 = np.float64(x)
    y_f64 = np.float64(y)
    
    err = abs(x_f64 - y_f64)
    
    return err
```

```python
# Ej 2

def error_relativo(x, y):
    x_f64 = np.float64(x)
    y_f64 = np.float64(y)
    
    if x_f64 == 0:
        raise ValueError("El valor de x no puede ser cero para calcular el error relativo.")
        
    return abs(x_f64 - y_f64) / abs(x_f64)
```

```python
def norma1(A):
    A = np.array(A)
    n,m = np.shape(A)
    normas = np.zeros(m)  
    for j in range(m):
        suma_columna = 0
        for i in range(n):
            suma_columna += abs(A[i, j])
        normas[j] = suma_columna
            
    return np.max(normas)

def normaliza(X, p):
    X_norm = []
    for v in X:
        v_arr = np.array(v, dtype=float)
        X_norm.append(v_arr / norma(v_arr, p))
    return X_norm

def normaInf(A):
    A = np.array(A)
    n,m = np.shape(A)
    normas = np.zeros(n)  
    for i in range(n):
        suma_fila = 0
        for j in range(m):
            suma_fila += abs(A[i, j])
        normas[i] = suma_fila
            
    return np.max(normas)

def normaExacta(A, p=[1,'inf']):
    if p == 1:
        return norma1(A)
    elif p == 'inf':
        return normaInf(A)
    elif p == [1,'inf']:
        return [norma1(A), normaInf(A)]
    else:
        return None

def norma(x, p):
    x_len = len(x)
    
    if p == 'inf':
        maximo = abs(x[0])
        for i in range(1, x_len):
            if abs(x[i]) > maximo:
                maximo = abs(x[i])
        return maximo
    
    norma = 0
    for i in range(x_len):
        norma += abs(x[i])**p
    
    return norma**(1/p)


def normaMatMC(A, q, p, Np):
    A = np.array(A)
    n, m = np.shape(A)
    
    mejor_valor = 0
    mejor_x = None
    
    for _ in range(Np):
        v = np.random.rand(m) * 2 - 1 
        x = v / norma(v, p)             
        
        y = A@x                    
        valor = norma(y, q)            
        
        if valor > mejor_valor:
            mejor_valor = valor
            mejor_x = x
    
    return mejor_valor, mejor_x



def condMC(A, p):
    Np=10000
    normaA, _ = normaMatMC(A, p, p, Np)
    A_inv = np.linalg.inv(A)
    normaAinv, _ = normaMatMC(A_inv, p, p, Np)
    
    return normaA * normaAinv
```
